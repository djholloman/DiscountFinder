### 1. Setup Gemini API Client
To search the internet using Gemini, we must configure the client. Make sure you have added your `GOOGLE_API_KEY` to Colab's secrets (🔑 icon on the left panel).

```python
from google import genai
from google.colab import userdata
from google.genai import types

# Retrieve API key and initialize client
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
client = genai.Client(api_key=GOOGLE_API_KEY)
```

### 2. Search for Discounted Shopping Items
We will define a function that queries Gemini using Google Search grounding enabled. This allows the model to fetch real-time search results to find active sales and discounted products.

```python
def search_discounted_items(query: str):
    prompt = f"Search the web and list 5-10 specific shopping items on sale or discounted matching this query: '{query}'. Provide the product name, current sale price, original price (if available), store, and a brief description or link if found."

    response = client.models.generate_content(
        model='gemini-2.5-flash',
        contents=prompt,
        config=types.GenerateContentConfig(
            tools=[types.Tool(google_search=types.GoogleSearch())]
        )
    )
    return response.text

# Example search: Electronics on sale
results = search_discounted_items("electronics headphones on sale")
print(results)
```
